## Single Index Model

This is an experiment to teach myself about the single index model. At it's core, it models each stock's returns as:

r_i = alpha_i + (beta_i * r_SPY) + epsilon_i

The key finding is that stock-specific risk shrinks with 1/N, while market risk has a risk floor we cant go past. This makes sense; we are always at risk of some downturn or noise that affects the general market.

## Notes:
- we want alpha to be positive, it is the y-intercept on the chart of stock returns vs market returns.
- Beta is the sensitivity / multiplier of the stock
- epsilon is the residual, or the amount of return which was explained by individual changes to the stock
